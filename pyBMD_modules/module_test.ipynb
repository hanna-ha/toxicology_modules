```python
import os
from typing import List, Optional
 
_CLEAN_EXPORT_COLS = [
    "gene",
    "model_name",
    "recommended",
    "bmd",
    "bmdl",
    "bmdu",
    "aic",
    "p_value1",     # Test 1 (mean vs control)
    "p_value2",     # Test 2 (constant variance)
    "p_value3",     # Test 3 (modelled variance)
    "p_value4",     # Test 4 (goodness-of-fit)
    "dof",
    "chi_squared",
    "bmr",
    "bmr_type",
    "num_params",
    "loglikelihood",
]
 
 
def clean_results_for_export(
    results: pd.DataFrame,
    extra_cols: Optional[List[str]] = None,
) -> pd.DataFrame:

    keep = list(_CLEAN_EXPORT_COLS)
    if extra_cols:
        keep.extend(extra_cols)
 
    # Only keep columns that actually exist in the DataFrame
    keep = [c for c in keep if c in results.columns]
 
    return results[keep].copy()
 
 
def export_results(
    results: pd.DataFrame,
    path: str,
    clean: bool = True,
    extra_cols: Optional[List[str]] = None,
) -> None:
    """

    """
    if clean:
        out = clean_results_for_export(results, extra_cols=extra_cols)
    else:
        out = results
 
    ext = os.path.splitext(path)[1].lower()
    if ext in (".xlsx", ".xls"):
        out.to_excel(path, index=False)
    else:
        out.to_csv(path, index=False)
 
    print(f"Saved {len(out)} rows × {len(out.columns)} cols → {path}")
```
